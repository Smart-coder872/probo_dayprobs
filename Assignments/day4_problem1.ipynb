```python
x = array([[0, 2], [1, 1], [2, 0]])
xT = x.T
```

```python
def covariance(X):
    x_x = []
    x_y = []
    y_x = []
    y_y = []
    
    n = len(X.T)
    for i in X:
        x_all = X[0]
        y_all = X[1]
    for i in x_all:
       x_part = x_all[i] - mean(x_all)
       y_part = y_all[i] - mean(y_all)
       
       x_x.append(x_part*x_part)
       x_y.append(x_part*y_part)
       y_x.append(y_part*x_part)
       y_y.append(y_part*y_part)
    
    result_xx = (1/(n-1))*sum(x_x)
    result_xy = (1/(n-1))*sum(x_y)
    result_yx = (1/(n-1))*sum(y_x)
    result_yy = (1/(n-1))*sum(y_y)

    result = [[result_xx, result_xy], [result_yx, result_yy]]
    return result
covariance(x.T)
```
